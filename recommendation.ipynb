```python
import pandas as pd

#ดึงไฟล์เสื้อผ้า.csvมาอ่านก่อน
df = pd.read_csv('myntra_products_catalog.csv')

#อ่านไฟล์ข้อมูล 5 แถวแรกของไฟล์
df.head()
```

```python
print("All column:")
#บอกชื่อคอลัมน์ทั้งหมดในไฟล์
print(df.columns.tolist())

#บอกว่าไฟล์นี้มีข้อมูบอะไรบ้าง ของเเต่ละคอลัมน์
df.info()
```

```python
#การจัดการกับช่องที่ไม่มีข้อมูลในเเต่ละคอลัมน์ โดยการเติมเป็นค่าว่าง เเละยังสร้างคอลัมน์ใหม่ที่ชื่อว่า tags
df['tags'] = df['ProductName'].fillna('') + ' ' + df['ProductBrand'].fillna('') + ' ' + df['PrimaryColor'].fillna('')

print(df['tags'].head())
```

```python
#เรียกlibraryมาใช้
from sklearn.feature_extraction.text import TfidfVectorizer

#สร้างตัวเเปลให้ stop_words='english' ช่วยตัดคำเชื่อมออก เช่น a, an, the
tfidf = TfidfVectorizer(stop_words='english')

#สร้างตัวเเปรเเละใช้ .fit_transform ในการบอกจำนวนROWเเละCOLUMNใน tags
tfidf_matrix = tfidf.fit_transform(df['tags'])

#ปริ้นจำนวน(rowทั้งหมด เสื้อผ้า, columnทั้งหมด คำศัพท์ที่ไม่ซ้ำกัน ตัดคำซ้ำออกเเล้ว)
print("Matrix shape:",tfidf_matrix.shape)
```
